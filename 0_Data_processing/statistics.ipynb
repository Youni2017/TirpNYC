```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path(".")

files = {
    "yellow": DATA_DIR / "yellow_cleaned_df.csv",
    "green":  DATA_DIR / "green_cleaned_df.csv",
    "fhv":    DATA_DIR / "fhv_cleaned_df.csv",
}

for name, path in files.items():
    print("=" * 70)
    print(f"Descriptive Statistics for: {name}")
    print("=" * 70)

    df = pd.read_csv(path)


    print(df.describe().T)    
    print("\n\n")
```
