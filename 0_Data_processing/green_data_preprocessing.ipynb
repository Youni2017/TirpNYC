```python
import pandas as pd

green_df = pd.read_csv("green_tripdata_2025-08.csv")

print(green_df.shape)     
print(green_df.columns)    
print(green_df.head())   
```

```python
green_df.head()
```

```python
green_df = green_df[["lpep_pickup_datetime","lpep_dropoff_datetime","PULocationID","DOLocationID","total_amount"]]
green_df
```

```python
green_df.isna().sum()
```

```python
green_df["lpep_pickup_datetime"] = pd.to_datetime(green_df["lpep_pickup_datetime"], errors="coerce")
green_df["lpep_dropoff_datetime"] = pd.to_datetime(green_df["lpep_dropoff_datetime"], errors="coerce")


green_df[["lpep_pickup_datetime", "lpep_dropoff_datetime"]].isna().sum()
```

```python

invalid_time = green_df[green_df["lpep_dropoff_datetime"] <= green_df["lpep_pickup_datetime"]]
len(invalid_time)  
```

```python
green_df = green_df[green_df["lpep_dropoff_datetime"] > green_df["lpep_pickup_datetime"]]
```

```python
green_df["trip_minutes"] = (
    (green_df["lpep_dropoff_datetime"] - green_df["lpep_pickup_datetime"]).dt.total_seconds() / 60
)


green_df[(green_df["trip_minutes"] < 5) & (green_df["trip_minutes"] > 180)]
```

```python
green_df = green_df[(green_df["trip_minutes"] >= 5) | (green_df["trip_minutes"] < 180)]
green_df
```

```python
green_df = green_df[green_df["total_amount"] > 0]
green_df
```

```python
green_df[["PULocationID", "DOLocationID"]].describe()

invalid_loc = green_df[
    (green_df["PULocationID"] <= 0) |
    (green_df["DOLocationID"] <= 0) |
    (green_df["PULocationID"] > 263) |
    (green_df["DOLocationID"] > 263)
]
len(invalid_loc), invalid_loc.head()
```

```python
green_df = green_df[(green_df['PULocationID']<=263) & (green_df['PULocationID']>=1)]
green_df = green_df[(green_df['DOLocationID']<=263) & (green_df['DOLocationID']>=1)]
```

```python
green_df = green_df.rename(columns={'lpep_pickup_datetime': 'pickup_datetime', 'lpep_dropoff_datetime': 'dropoff_datetime', 'PULocationID': 'pickup_location', 'DOLocationID': 'dropoff_location '})
```

```python
green_df = green_df.drop(columns=["trip_minutes"])
```
